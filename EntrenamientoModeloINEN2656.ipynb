```python
# ✅ Paso 4: Entrenar el modelo YOLOv11 desde cero
model = YOLO("yolo11m.pt")  # Puedes usar yolo11n.pt (ligero), yolo11m.pt (más potente)

results = model.train(
    data=os.path.join(dataset.location, "data.yaml"),
    epochs=50,
    imgsz=960,
    batch=8,
    name="roboflow_yolov11_model",
    device=0 if torch.cuda.is_available() else 'cpu',
    patience=20,
    verbose=True
)
```
